in this code I take BART_Subject_demographics.csv and change some columns and add more columns to it for SOZ information

```python
import pandas as pd
from pathlib import Path
```

```python
file_path = Path(
    r"D:\Nill\code\BART\IED\0_0_new_IED_new_area_labels\0_demographic"
) / "BART_Subject_demographics.csv"
```

```python
df = pd.read_csv(file_path)

print(df.head())
```

```python
df.rename(columns={"age": "age_at_testing"}, inplace=True)
print(df.head())
```

```python
# Add new empty columns
new_columns = [
    "DOB",
    "Epilepsy_type",
    "SOZ_onset_area",
    "SOZ_onset_chans",
    "SOZ_recruit_area",
    "SOZ_recruit_chans"
]

for column in new_columns:
    df[column] = pd.NA
```

```python
output_path = file_path.with_name("BART_Subject_demographics_SOZ.csv")
df.to_csv(output_path, index=False)
```
